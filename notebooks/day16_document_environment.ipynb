## 1. Record the packages needed by these notebooks
`requirements.txt` lists installable packages, using `package==version` to record the environment
that produced your results. Python itself is recorded separately in `environment_info.json`.
Standard-library modules such as `pathlib`, `json`, `ast`, and `sys` do not belong in requirements.

The core list covers the Day 13–16 notebooks and their shared modules. If `ipykernel` is installed,
its version is also recorded. You can use your existing Jupyter or VS Code application.

The saved-CSV analysis does not need to download data again. If you also want to document the earlier
FRED download cell using pandas-datareader, set `INCLUDE_FRED_DOWNLOAD_PACKAGE = True` below.
That optional package must already be installed to record its version.

````python
INCLUDE_FRED_DOWNLOAD_PACKAGE = False
packages = ['numpy', 'pandas', 'matplotlib', 'scikit-learn', 'joblib']
try:
    version('ipykernel')
    packages.append('ipykernel')
except PackageNotFoundError:
    pass
if INCLUDE_FRED_DOWNLOAD_PACKAGE:
    packages.append('pandas-datareader')

package_versions = {}
missing = []
for package in packages:
    try:
        package_versions[package] = version(package)
    except PackageNotFoundError:
        missing.append(package)
assert not missing, (
    f'Missing packages in this kernel: {missing}. Use the same environment as your completed notebooks. '
    'If you only analyze the saved CSV, leave INCLUDE_FRED_DOWNLOAD_PACKAGE=False.'
)
requirements_text = ''.join(f'{name}=={package_versions[name]}\n' for name in packages)
environment_info = {
    'python_version': platform.python_version(),
    'python_implementation': platform.python_implementation(),
    'platform': platform.platform(),
    'packages': package_versions,
    'scope': 'Day 13–16 finance forecasting notebooks and shared modules',
}
print(requirements_text)
print('Python version is recorded separately:', environment_info['python_version'])
````

## 2. Save the environment files
This writes `requirements.txt` and `environment_info.json` in the project root.
If either file already contains different text, it creates a numbered backup before writing the new
version. Running this cell again without changing versions does not create another backup.
These two files document direct dependencies and Python/platform details. They are not a complete
Conda lockfile or a guarantee that every operating system will produce identical numerical results.

````python
def write_with_backup(path, text):
    """Save text, retaining a numbered copy of a different existing file."""
    if path.exists():
        if path.read_text(encoding='utf-8') == text:
            print('Already current:', path.name)
            return
        number = 1
        backup = path.with_name(f'{path.name}.backup{number}')
        while backup.exists():
            number += 1
            backup = path.with_name(f'{path.name}.backup{number}')
        backup.write_bytes(path.read_bytes())
        print('Previous file preserved:', backup.name)
    path.write_text(text, encoding='utf-8')
    print('Saved:', path)

write_with_backup(PROJECT_ROOT / 'requirements.txt', requirements_text)
write_with_backup(PROJECT_ROOT / 'environment_info.json', json.dumps(environment_info, indent=2) + '\n')
````
